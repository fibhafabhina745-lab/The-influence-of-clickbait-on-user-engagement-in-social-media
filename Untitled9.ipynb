```python

from google.colab import files
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Upload your Excel dataset
uploaded = files.upload()
file_name = next(iter(uploaded))

# Read the Excel file
df = pd.read_excel(file_name, sheet_name="export")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
display(df.head())
```

```python

# Dataset dimensions
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

# Column names and data types
print("\nColumn names and data types:")
display(pd.DataFrame({
    "Column": df.columns,
    "Data type": df.dtypes.astype(str).values
}))

# Dataset information
print("\nDataset information:")
df.info()

# Duplicate rows
print("\nExact duplicate rows:", df.duplicated().sum())

# Country distribution
print("\nPosts by country:")
display(df["UK_USA"].value_counts(dropna=False))

# Check blank strings in text columns
print("\nBlank strings in text columns:")
for col in df.select_dtypes(include=["object", "string"]).columns:
    blanks = df[col].astype("string").str.strip().eq("").sum()
    print(f"{col}: {blanks}")
```

```python

# Count missing values in every column
missing = df.isna().sum()

missing_audit = pd.DataFrame({
    "Missing values": missing,
    "Missing percentage": (missing / len(df) * 100).round(2)
})

missing_audit = missing_audit.sort_values(
    "Missing values", ascending=False
)

print("Columns containing missing values:")
display(missing_audit[missing_audit["Missing values"] > 0])

print("\nTotal missing cells:", int(df.isna().sum().sum()))
print("Total cells in dataset:", df.size)

# Check for unusual negative values in numeric columns.
# Negative residuals may be valid, so these are flagged for inspection,
# not automatically treated as missing values.
print("\nNumeric columns containing negative values:")
for col in df.select_dtypes(include="number").columns:
    count = (df[col] < 0).sum()
    if count > 0:
        print(f"{col}: {count} negative values")
```

```python

# Summary statistics for engagement variables
engagement_cols = [
    "post_likes_count",
    "post_shares_count",
    "post_allReactions"
]

display(df[engagement_cols].describe())

# Headline word count summary
display(df.groupby("UK_USA")["headline_word_count"].describe())

# Check whether total reactions equal the listed reaction categories
reaction_cols = [
    "post_reactions_love_count",
    "post_reactions_haha_count",
    "post_reactions_wow_count",
    "post_reactions_sad_count",
    "post_reactions_angry_count"
]

reaction_sum = df[reaction_cols].sum(axis=1, min_count=len(reaction_cols))
valid = df["post_allReactions"].notna() & reaction_sum.notna()

print(
    "Rows checked for reaction consistency:",
    int(valid.sum())
)
print(
    "Rows where total differs from listed reaction categories:",
    int((df.loc[valid, "post_allReactions"] != reaction_sum[valid]).sum())
)
```

```python

shares = df["post_shares_count"].dropna()
positive_shares = shares[shares > 0]

plt.figure(figsize=(8, 5))

bins = np.logspace(
    np.log10(positive_shares.min()),
    np.log10(positive_shares.max()),
    30
)

plt.hist(positive_shares, bins=bins, edgecolor="black")
plt.xscale("log")
plt.xlabel("Number of shares (log scale)")
plt.ylabel("Number of posts")
plt.title("Distribution of Facebook Post Shares")
plt.tight_layout()
plt.show()

print("Posts with zero shares:", int((shares == 0).sum()))
```

```python

plot_data = df[
    ["post_likes_count", "post_shares_count"]
].dropna()

# Log scales cannot display zero or negative values
plot_data = plot_data[
    (plot_data["post_likes_count"] > 0) &
    (plot_data["post_shares_count"] > 0)
]

plt.figure(figsize=(8, 5))
plt.scatter(
    plot_data["post_likes_count"],
    plot_data["post_shares_count"],
    alpha=0.4
)

plt.xscale("log")
plt.yscale("log")
plt.xlabel("Number of likes (log scale)")
plt.ylabel("Number of shares (log scale)")
plt.title("Relationship Between Likes and Shares")
plt.tight_layout()
plt.show()

print("Rows included in scatter plot:", len(plot_data))
print(
    "Pearson correlation:",
    round(
        plot_data["post_likes_count"].corr(
            plot_data["post_shares_count"]
        ), 3
    )
)
```

```python

plot_data = df[["UK_USA", "headline_word_count"]].dropna()

countries = sorted(plot_data["UK_USA"].unique())

groups = [
    plot_data.loc[
        plot_data["UK_USA"] == country,
        "headline_word_count"
    ]
    for country in countries
]

plt.figure(figsize=(8, 5))
plt.boxplot(groups, tick_labels=countries, showfliers=False)
plt.xlabel("Country group")
plt.ylabel("Headline word count")
plt.title("Headline Length by Country")
plt.tight_layout()
plt.show()

print("Posts included in box plot:", len(plot_data))
```

```python

# Check whether post_likes_count is included in the reaction total

reaction_cols = [
    "post_reactions_love_count",
    "post_reactions_haha_count",
    "post_reactions_wow_count",
    "post_reactions_sad_count",
    "post_reactions_angry_count"
]

listed_reactions = df[reaction_cols].sum(axis=1)

# Test 1: sum of the five listed categories
print(
    "Matches without likes:",
    (df["post_allReactions"] == listed_reactions).sum()
)

# Test 2: sum of likes plus the five listed categories
listed_plus_likes = listed_reactions + df["post_likes_count"]

print(
    "Matches with likes added:",
    (df["post_allReactions"] == listed_plus_likes).sum()
)

# Display sample rows for investigation
display(df[[
    "post_likes_count",
    *reaction_cols,
    "post_allReactions"
]].head(10))
```

```python
# Check common special missing-value codes
numeric_df = df.select_dtypes(include="number")

for code in [-999, -99, 999999]:
    matches = (numeric_df == code).sum()
    matches = matches[matches > 0]

    print(f"\nOccurrences of {code}:")
    print(matches if not matches.empty else "None found")
```
